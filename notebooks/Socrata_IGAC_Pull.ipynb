```python
import pandas as pd
from sodapy import Socrata

client = Socrata("www.datos.gov.co", None)
registros = []
lote = 50000
offset = 0

while True:
    datos = client.get("7y2j-43cv", where="departamento = 'BOGOTÁ D.C.' and municipio = 'BOGOTÁ D.C.'", limit=lote, offset=offset)
    if not datos:
        break
    registros.extend(datos)
    offset += lote

df = pd.DataFrame.from_records(registros)
df.to_csv("IGAC.csv", index=False)
print(f"Total registros guardados: {len(df)}")
```
